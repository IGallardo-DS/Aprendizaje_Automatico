# 1. Cargar el conjunto de datos en una estructura adecuada para su manipulación.

```python
import pandas as pd
df_automobile = pd.read_csv('Archivos/Automobile.csv')
df_automobile.head()
```

# 2. Crear un gráfico de dispersión para visualizar las relaciones entre los pares de características numéricas `mpg` y `weight`.

```python
import matplotlib.pyplot as plt
plt.scatter(df_automobile["weight"], df_automobile["mpg"])
plt.xlabel("Peso")
plt.ylabel("Rendimiento Combustible")
plt.title("Relación Peso y Combustible")
plt.show()
```

# 3. Generar un histograma del consumo de combustible (`mpg`) para analizar su distribución.

```python
df_automobile["mpg"].hist(bins=15, grid=False)
plt.title("Histograma de mpg")
plt.xlabel("mpg")
plt.ylabel("Frecuencia")
plt.show()
```

# 4. Graficar un diagrama de caja (boxplot) de las características `cylinders` y `origin` para identificar posibles valores atípicos.

```python
df_automobile.boxplot(column="cylinders", by="origin")
plt.title("Distribución de Cilindros por Origen")
plt.suptitle("")  # Quita el subtítulo automático que genera pandas ("Boxplot grouped by origin")
plt.xlabel("Origen")
plt.ylabel("Cilindros")
plt.show()
```

# 5. Realizar un gráfico de barras que muestre la distribución de automóviles según su año de modelo (`model_year`) y origen (`origin`).

```python
# Contar la cantidad de registros por año y por origen
df_automobile.groupby(["model_year", "origin"]).size().unstack().plot(kind="bar")

plt.title("Cantidad de autos por Año y Origen")
plt.xlabel("Año del modelo")
plt.ylabel("Cantidad")
plt.show()
```
